### Question 2


Input layer → 3 inputs

Hidden layer → 2 neurons

Output layer → 2 neurons

Activation: Sigmoid

Loss function: Mean Squared Error (MSE)

```python
import numpy as np
import pandas as pd
```

# Activation Functions

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def delta_sigmoid(z):
    s = sigmoid(z)
    return s * (1 - s)

def step(z):
    return (z >= 0.5).astype(int)
```

# Feedforward and Backpropagation


```python
class Layer:
    def __init__(self, W):
        self.W = W

    def forward(self, X):
        self.X = X                    
        self.Z = X @ self.W.T         
        self.A = sigmoid(self.Z)      
        return self.A

    def backward(self, dA, alpha):
        dZ = dA * delta_sigmoid(self.Z)   
        dW = dZ.T @ self.X                
        dX = dZ @ self.W                 

        self.W -= alpha * dW              
        return dX
```

# Inputs given 

```python
X = np.array([[1, 0.7, 1.2]])
y = np.array([[1, 0]])

W1 = np.array([[0.5, 1.5, 0.8],
               [0.8, 0.2, -1.6]])

W2 = np.array([[0.9, -1.7, 1.6],
               [1.2,  2.1, -0.2]])

hidden_layer = Layer(W1)
output_layer = Layer(W2)
```

# Training parameters

```python
epochs = 1000
alpha = 0.1
theta = 0.5
```

# Training

```python
for epoch in range(epochs):

    h = hidden_layer.forward(X)
    h_bias = np.concatenate([np.ones((1,1)), h], axis=1)
    y_pred = output_layer.forward(h_bias)

    loss = 0.5 * np.sum((y - y_pred) ** 2)

    dL_dy = y_pred - y
    d_hidden_bias = output_layer.backward(dL_dy, alpha)
    d_hidden = d_hidden_bias[:, 1:]     # Remove bias gradient
    hidden_layer.backward(d_hidden, alpha)

    if epoch % 100 == 0:
        print(f"Epoch {epoch} | Prediction: {y_pred} | Loss: {loss:.6f}")
```

# output


```python
print("Final continuous output:", y_pred)
print("Final thresholded output:", step(y_pred))
```
